# Progressive GAN growth

Reads `reports/mnist_pg_gan/generations/cifar/` after `scripts/mnist_pg_gan/train_progan.py`. The timeline is one row per phase. Resolution doubles from 4 to 32, and each new block is faded in before the stable phase at that size.


```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
from IPython.display import Image, display

archive = Path("../../reports/mnist_pg_gan/generations/cifar")
if not (archive / "growth.json").is_file():
    print("No archive yet. Run: uv run python scripts/mnist_pg_gan/train_progan.py")
else:
    growth = json.loads((archive / "growth.json").read_text(encoding="utf-8"))
    history = json.loads((archive / "history.json").read_text(encoding="utf-8"))
    fig, ax = plt.subplots()
    ax.plot(history["train_discriminator"], label="train discriminator")
    ax.plot(history["train_generator"], label="train generator")
    ax.set_xlabel("epoch")
    ax.set_ylabel("bce")
    twin = ax.twinx()
    twin.step(range(len(history["resolution"])), history["resolution"], where="post", color="0.35")
    twin.set_ylabel("resolution")
    ax.legend()
    plt.show()
    display(Image(filename=str(archive / "growth_timeline.png")))
    for frame in growth["frames"]:
        print(
            f"epoch {frame['epoch']:04d}  {frame['resolution']}px  {frame['phase']}  alpha={frame['alpha']}"
        )
        display(Image(filename=str(archive / frame["file"])))
```
